```python
import math

# Generating sqaure root of the given number

# By definition, the square root of a number is a value that, when multiplied by itself, gives the original number ($\sqrt{n} \times \sqrt{n} = n$). Since $n^{0.5} \times n^{0.5}$ also equals $n$, then $n^{0.5}$ must be the square root!
def generateSqrt(num):
    return num**0.5

print(generateSqrt(16))
```

```python
# Find the Largest number in the given array
arr = [1,2,3,4,52,1,3,4,5]

def findLargest(a):
    largest = a[0]

    for i in range(len(a)):
        if(largest < a[i]):
            largest = a[i]

    return largest

print(findLargest(arr))
```

```python
# Find the Second Largest number in the given array

import math

arr = [1,2,33,2,1,34,21,24,553,-2, 553]

largest = -math.inf
second_largest = -math.inf

for elem in arr:
    if elem > largest:
        second_largest = largest 
        largest = elem
    elif elem > second_largest and elem != largest:
        second_largest = elem

print(largest, second_largest)
```

```python
# Checking if the string is palindrome or not

str = "maam"

def is_palindrome(s):

    left = 0
    right = len(s)-1

    while left < right:

        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True


print(is_palindrome(str))

# print(str == str[::-1])
```

```python
# Find if 2 dictionary is equal 

dict1 = {
  "a":1,
  "b":2
}
dict2 = {
  "a":1,
  "b":2
}

def dict_equal(d1, d2):
    # Checking if the length of the dict is equal
    if len(d1) != len(d2):
        return False
    
    for key in d1:
        # check if both the dicts have the same keys
        if key not in d2:
            return False
        
        # As keys are equal then we can check whether values are equal or not using the keys
        if d1[key] != d2[key]:
            return False
    
    return True

print(dict_equal(dict1, dict2))
```

```python
# 7. Write a program that will tell whether the given year is a leap year or not

# Leap Year Rules

# A year is a leap year if:

# It is divisible by 400, OR
# It is divisible by 4 and not divisible by 100.

year = int(input("Enter the year to check whether it is leap year or not"))

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print("Leap year")
else:
    print("Not a Leap Year")
```
