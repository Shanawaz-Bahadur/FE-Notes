```python
import math

# Generating sqaure root of the given number

# By definition, the square root of a number is a value that, when multiplied by itself, gives the original number ($\sqrt{n} \times \sqrt{n} = n$). Since $n^{0.5} \times n^{0.5}$ also equals $n$, then $n^{0.5}$ must be the square root!
def generateSqrt(num):
    return num**0.5

print(generateSqrt(16))
```

```python
# Find the Largest number in the given array
arr = [1,2,3,4,52,1,3,4,5]

def findLargest(a):
    largest = a[0]

    for i in range(len(a)):
        if(largest < a[i]):
            largest = a[i]

    return largest

print(findLargest(arr))
```

```python
# Find the Second Largest number in the given array

import math

arr = [1,2,33,2,1,34,21,24,553,-2, 553]

largest = -math.inf
second_largest = -math.inf

for elem in arr:
    if elem > largest:
        second_largest = largest 
        largest = elem
    elif elem > second_largest and elem != largest:
        second_largest = elem

print(largest, second_largest)
```

```python
# Reversing the string

str = "Hello"

# Method 1: Revesing the string and storing it in other variable
rev = ""
for i in range(len(str)-1, -1, -1):
    rev += str[i]

print(rev)

# Methos 2: revering the existing string
# print(str[::-1])
```

```python
# Checking if the string is palindrome or not

str = "maam"

def is_palindrome(s):

    left = 0
    right = len(s)-1

    while left < right:

        if s[left] != s[right]:
            return False

        left += 1
        right -= 1

    return True


print(is_palindrome(str))

# print(str == str[::-1])
```

```python
# Character counting 

str = "banana"
dict = {}

for ch in str:
    if ch not in dict.keys():
        dict[ch] = 1
    else:
        dict[ch] += 1

print(dict)
```
